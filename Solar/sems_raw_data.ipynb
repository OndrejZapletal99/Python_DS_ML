```python
import pandas as pd
import numpy as np
import sys
import os
```

```python
print(os.getcwd())
```

```python

xls_folder = './Raw Data/SEMS'


xls_files = [f for f in os.listdir(xls_folder) if f.endswith('.xls')]


for f in xls_files:
    print(f)
```

```python
df_raw = pd.DataFrame()
```

```python

for file in xls_files:
    full_path = os.path.join(xls_folder, file)
    df_temp = pd.read_excel(full_path)
    df_temp['Unnamed: 0'] = df_temp['Unnamed: 0'].astype(str)
    df_temp = df_temp[~df_temp['Unnamed: 0'].str.lower().isin(['nan', 'měsíční zpráva', 'total'])]
    df_temp
    df_temp.columns = df_temp.iloc[0]
    df_temp = df_temp[1:].reset_index(drop = True)
    
    df_raw = pd.concat([df_raw, df_temp], ignore_index=True)
```

```python
df_raw['Datum'] = pd.to_datetime(df_raw['Datum'], errors='coerce', dayfirst=True)

# Add columns for month and year
df_raw['Month'] = df_raw['Datum'].dt.month
df_raw['Year'] = df_raw['Datum'].dt.year

# Check number of records in each year
df_check = df_raw['Month'].value_counts().sort_index()
df_check
```

```python
pivot = pd.pivot_table(
    df_raw,
    index='Year',
    columns='Month',
    values='Datum',
    aggfunc='count',
    fill_value=0
)
pivot
```

```python
df_raw.to_csv(r'.\Data\SEMS_data.csv', index=False)
```
