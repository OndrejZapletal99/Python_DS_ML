```python
import requests
import pandas as pd

# --- URL s klíčem (můžeš ho načíst z txt podle dřívějšího postupu) ---
url = "https://weather.visualcrossing.com/VisualCrossingWebServices/rest/services/timeline/Vala%C5%A1sk%C3%A9%20Mezi%C5%99%C3%AD%C4%8D%C3%AD%2C%20Zl%C3%ADnsk%C3%BD%20kraj%2C%20%C4%8Cesk%C3%A1%20Republika?unitGroup=metric&include=days&key=H5X3D8N2WKAAQ68QDPHD5UA8J&contentType=json"

# --- Stažení dat ---
response = requests.get(url)
data = response.json()
```

```python
# --- Získání seznamu dnů ---
days = data.get("days", [])

# --- Převod na DataFrame ---
df = pd.json_normalize(days)

# --- Zobrazení prvních 5 řádků ---
df.head()
```

```python
df.to_csv("Data/weather_data_forecast.csv", index=False)
```
