```python
def dataframe_info(df: pd.DataFrame):
    """
    Generate a summary DataFrame containing metadata about the columns of the input DataFrame.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        pd.DataFrame: A summary DataFrame with the following columns:
            - Column_name: Name of each column.
            - Total records: Total number of records in each column.
            - Missing Values: Number of missing (NaN) values in each column.
            - Data type: Data type of each column.
            - Unique values: Number of unique values in each column.
    """
    df_summary = pd.DataFrame({
        'Column_name': df.columns,
        'Total records': [df[col].size for col in df.columns],
        'Missing Values': [df[col].isna().sum() for col in df.columns],
        'Data type': [df[col].dtype for col in df.columns],
        'Unique values': [df[col].nunique() for col in df.columns]
    })

    return df_summary
```

# 2. Feature Engineering

## 2.1 SEMS Solar Data

```python
df_solar.columns
```

```python
df_solar_data = df_solar[['Datum','PV(kWh)']].copy()
```

```python
df_solar_data
```

```python
df_solar_data.dtypes
```

```python
df_solar_data['Datum'] = df_solar_data['Datum'].str.replace(r'\.', '-', regex=True)
df_solar_data['Datum'] = pd.to_datetime(df_solar_data['Datum'], dayfirst=True)
df_solar_data
```

```python
df_solar_data['month'] = df_solar_data['Datum'].dt.month
df_solar_data['dayofyear'] = df_solar_data['Datum'].dt.dayofyear

df_solar_data['sin_month'] = np.sin(2 * np.pi * df_solar_data['month'] / 12)
df_solar_data['cos_month'] = np.cos(2 * np.pi * df_solar_data['month'] / 12)


df_solar_data['sin_dayofyear'] = np.sin(2 * np.pi * df_solar_data['dayofyear'] / 365)
df_solar_data['cos_dayofyear'] = np.cos(2 * np.pi * df_solar_data['dayofyear'] / 365)

df_solar_data['year'] = df_solar_data['Datum'].dt.year

```

```python
df_solar_data.head(5)
```

```python
df_solar_data_final = df_solar_data.drop(columns=['month', 'dayofyear'])
```

```python
df_solar_data_final.head(5)
```

## 2.2 Weather Data

```python
df_weather_data = df_weather[list(common_columns)].copy()
df_weather_data.head(5)
```

```python
df_weather_data.select_dtypes('object')
```

```python
df_weather_data['datetime'] = pd.to_datetime(df_weather_data['datetime'])
```

```python
df_weather_data['month'] = df_weather_data['datetime'].dt.month
df_weather_data['dayofyear'] = df_weather_data['datetime'].dt.dayofyear

df_weather_data['sin_month'] = np.sin(2 * np.pi * df_weather_data['month'] / 12)
df_weather_data['cos_month'] = np.cos(2 * np.pi * df_weather_data['month'] / 12)


df_weather_data['sin_dayofyear'] = np.sin(2 * np.pi * df_weather_data['dayofyear'] / 365)
df_weather_data['cos_dayofyear'] = np.cos(2 * np.pi * df_weather_data['dayofyear'] / 365)

df_weather_data['year'] = df_weather_data['datetime'].dt.year
```

```python
def time_to_seconds(t):
    """Převede časový string HH:MM:SS na počet sekund od půlnoci."""
    h, m, s = map(int, t.split(':'))
    return h*3600 + m*60 + s


df_weather_data['sunset_seconds'] = df_weather_data['sunset'].apply(time_to_seconds)
df_weather_data['sunrise_seconds'] = df_weather_data['sunrise'].apply(time_to_seconds)


seconds_in_day = 24*3600
df_weather_data['sunset_sin'] = np.sin(2 * np.pi * df_weather_data['sunset_seconds'] / seconds_in_day)
df_weather_data['sunset_cos'] = np.cos(2 * np.pi * df_weather_data['sunset_seconds'] / seconds_in_day)


df_weather_data['sunrise_sin'] = np.sin(2 * np.pi * df_weather_data['sunrise_seconds'] / seconds_in_day)
df_weather_data['sunrise_cos'] = np.cos(2 * np.pi * df_weather_data['sunrise_seconds'] / seconds_in_day)


df_weather_data['day_length_seconds'] = df_weather_data['sunset_seconds'] - df_weather_data['sunrise_seconds']


df_weather_data['day_length_hours'] = df_weather_data['day_length_seconds'] / 3600
```

```python
df_weather_data['preciptype']= df_weather_data['preciptype'].fillna("['none']")
df_weather_data['preciptype'].value_counts()
```

```python
df_weather_data.columns
```

```python
df_weather_data_final = df_weather_data.drop(columns=[ 'sunset', 'sunrise', 'month', 'dayofyear'])
```

```python
df_final = pd.merge(df_solar_data_final, df_weather_data_final, left_on='Datum',right_on=  'datetime', how='inner')
```
