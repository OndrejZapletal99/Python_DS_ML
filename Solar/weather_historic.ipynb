```python
import requests
import pandas as pd

# --- URL s klíčem (můžeš ho načíst z txt podle dřívějšího postupu) ---
url = "https://weather.visualcrossing.com/VisualCrossingWebServices/rest/services/timeline/Vala%C5%A1sk%C3%A9%20Mezi%C5%99%C3%AD%C4%8D%C3%AD/2023-09-01/2025-06-29?unitGroup=metric&elements=datetime%2Ctempmax%2Ctempmin%2Ctemp%2Cfeelslikemax%2Cfeelslikemin%2Cfeelslike%2Cdew%2Chumidity%2Cprecip%2Cprecipcover%2Cpreciptype%2Csnow%2Csnowdepth%2Cwindgust%2Cwindspeed%2Cwindspeedmax%2Cwindspeedmean%2Cwinddir%2Cpressure%2Ccloudcover%2Cvisibility%2Csolarradiation%2Csolarenergy%2Cuvindex%2Csevererisk%2Csunrise%2Csunset&include=obs%2Cremote%2Cstats%2Cdays&key=H5X3D8N2WKAAQ68QDPHD5UA8J&contentType=json"

# --- Stažení dat ---
response = requests.get(url)
data = response.json()

```

```python
# --- Získání seznamu dnů ---
days = data.get("days", [])

# --- Převod na DataFrame ---
df = pd.json_normalize(days)

# --- Zobrazení prvních 5 řádků ---
df.head()
```
